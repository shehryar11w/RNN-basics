```python
import torch
from torch.utils.data import Dataset
from torch.utils.data import DataLoader
```

```python
with open("data/tinyshakespeare/input.txt", "r", encoding="utf-8") as f:
    text = f.read()

split = int(0.9 * len(text))

train_text = text[:split]
val_text = text[split:]

print(len(train_text), len(val_text))
```

```python
chars = sorted(list(set(text)))
vocab_size = len(chars)

stoi = {ch:i for i,ch in enumerate(chars)}
itos = {i:ch for ch,i in stoi.items()}

print(vocab_size)
```

```python
from typing import Any


train_data = torch.tensor([stoi[c] for c in train_text], dtype=torch.long)
val_data   = torch.tensor([stoi[c] for c in val_text], dtype=torch.long)

class ShakespeareDataset(Dataset):
    def __init__(self, data, seq_len) -> None:
        self.data = data
        self.seq_len = seq_len

    def __len__(self):
        return len(self.data) - self.seq_len

    def __getitem__(self, index) -> Any:
        x = self.data[index:index + self.seq_len] # the current letter
        y = self.data[index+ 1: index + self.seq_len + 1] # the next letter

        return x, y


seq_len = 100
batch_size = 64

train_dataset = ShakespeareDataset(train_data, seq_len)
val_dataset = ShakespeareDataset(val_data, seq_len)

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=batch_size,
    shuffle=False
)
```

```python
import torch.nn as nn
import torch.nn.functional as F


class RNN(nn.Module):

    def __init__(self):
        super().__init__()
        self.tanh = nn.Tanh()
        self.Wx = nn.Parameter(torch.randn([100,65]))
        self.Wh = nn.Parameter(torch.randn([100,100]))
        self.Bx = nn.Parameter(torch.randn(100))
        self.Why = nn.Parameter(torch.randn(65, 100))
        self.By = nn.Parameter(torch.randn(65))

    def forward(self, x,h):
        outputs = []

        for t in range(x.shape[1]):
            char = x[:, t]
            char = F.one_hot(char, num_classes=65).float()
            h = self.tanh(char @ self.Wx.T + h @ self.Wh + self.Bx)
            logits = h @ self.Why.T + self.By
            outputs.append(logits)
            
        
        return torch.stack(outputs, dim=1), h


net = RNN()
```

```python
import torch.optim as optim

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(net.parameters(), lr=0.001, momentum=0.9)
```

```python
for epoch in range(15):  # loop over the dataset multiple times

    running_loss = 0.0
    
    for inputs, labels in train_loader:

        H = torch.zeros(inputs.size(0), 100)
        # zero the parameter gradients
        optimizer.zero_grad()

        # forward + backward + optimize
        outputs, H = net(inputs, H) 
        loss = criterion(
            outputs.reshape(-1, 65),
            labels.reshape(-1)
        )
        loss.backward()
        optimizer.step()

        H = H.detach()

        # print statistics
        running_loss += loss.item()
    train_loss = running_loss / len(train_loader)

    net.eval()
    val_loss = 0

    with torch.no_grad():
        for inputs, labels in val_loader:
            H = torch.zeros(inputs.size(0), 100)

            outputs, H = net(inputs, H)

            loss = criterion(
                outputs.reshape(-1, vocab_size),
                labels.reshape(-1)
            )

            val_loss += loss.item()

    val_loss /= len(val_loader)

    print(
        f"Epoch {epoch+1:2d} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f}"
    )

    net.train()

print('Finished Training')
```
